# **Klasifikasi Gambar Kucing vs. Anjing**

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

# Mengunduh dan memuat dataset CIFAR-10 (gunakan kelas burung dan kapal sebagai contoh dasar jika ingin cepat, atau dataset Kucing/Anjing)
# Untuk simplisitas gunakan CIFAR-10 yang menyediakan 10 kelas termasuk cat & dog
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()

# Normalisasi nilai piksel menjadi antara 0 dan 1
train_images, test_images = train_images / 255.0, test_images / 255.0
```

# **Model CNN**

```python
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10)
])

model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

# **Latih model**

```python
history = model.fit(train_images, train_labels, epochs=5,
                    validation_data=(test_images, test_labels))
```
